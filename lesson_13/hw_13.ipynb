## Постановка задачи

Необходимо реализовать мультимодального агента для редактирования изображений по текстовому запросу пользователя.

Агент должен:
- принимать на вход текстовую инструкцию пользователя;
- принимать на вход до 2 изображений;
- интерпретировать пользовательскую задачу;
- выбирать подходящие инструменты обработки изображений;
- выполнять редактирование входных фотографий;
- возвращать итоговый результат и подробные текстовые логи работы.

Примеры пользовательских запросов:
- «сделай фото как в журнале»;
- «добавь яркую желтую надпись сверху»;
- «убери такой-то объект с фото»;
- «сделай второе изображение визуально похожим на первое»;
- «выдели объект, затем затемни фон и добавь подпись».

## Что требуется реализовать

Нужно собрать агентную систему, в которой языковая или мультимодальная модель принимает решение о последовательности действий, а сами действия выполняются через набор инструментов.

Минимально в системе должны быть:
- агентный цикл;
- поддержка загрузки до 2 изображений;
- набор инструментов для редактирования изображений;
- механизм логирования шагов агента;
- сохранение и выдача итогового изображения пользователю.

## Требования к инструментам

Агент должен обладать набором инструментов редактирования изображений.

Обязательное требование:
- не менее 10 инструментов-фильтров или операций из числа рассмотренных на первой лекции.

Инструменты могут включать, например:
- изменение яркости;
- изменение контраста;
- изменение насыщенности;
- изменение резкости;
- размытие;
- гауссово размытие;
- медианный фильтр;
- изменение цветового тона;
- перевод в градации серого;
- бинаризацию;
- edge detection;
- морфологические операции;
- кроп;
- ресайз;
- поворот;
- наложение текста;
- маскирование области;
- инверсию цветов;
- денойзинг;
- эквализацию гистограммы.
- использование open-source diffusion-моделей для локального редактирования областей изображения;
- использование open-source моделей сегментации, например SAM, для поиска и выделения областей;
- выполнение произвольного Python-кода как отдельного инструмента, если он применяется к пользовательским изображениям;

Допускается реализовать больше инструментов, если это оправдано архитектурой решения.

## Дополнительные возможности

Дополнительный функционал не является обязательным, но допускается:
- вызов сторонних произвольных open-source моделей по API как tool для агента;

## Ограничения

Запрещено использовать API моделей, которые не выложены в open source.

Разрешено использовать (в том числе по любым доступным API):
- open-source LLM;
- open-source VLM;
- open-source diffusion-модели;
- open-source segmentation / detection / OCR / restoration модели;
- собственные локальные инструменты обработки изображений.

Если используется внешний API, необходимо, чтобы он соответствовал этому ограничению.

## Ожидаемое поведение агента

Агент должен уметь обрабатывать запросы разной степени определенности.

Система должна корректно работать как с:
- достаточно точными инструкциями;
- так и с более общими эстетическими запросами.

При этом от агента ожидается:
- разбор задачи пользователя;
- выбор релевантных инструментов;
- пошаговое применение инструментов;
- получение итогового визуально осмысленного результата.

## Логи

Результат работы агента должен обязательно содержать адекватные и удобные для чтения логи.

В логах должны быть:
- шаги рассуждения на уровне плана действий агента;
- информация о том, какой инструмент был вызван;
- аргументы вызова инструмента в текстовом виде;
- результат вызова инструмента;
- промежуточные решения агента;
- сообщения пользователю, если требуется уточнение или если задача не может быть выполнена напрямую.

Логи должны быть оформлены так, чтобы по ним можно было восстановить ход исполнения инструкции.

## Что должно поддерживаться на входе

На вход системе подаются:
- текстовый запрос пользователя;
- одно или два изображения.

Формат интерфейса выбирается самостоятельно. Это может быть, например:
- notebook;
- CLI;
- web-интерфейс;
- Telegram-бот;
- любой другой удобный формат.

## Что должно быть на выходе

На выходе система должна возвращать:
- итоговое отредактированное изображение или изображения;
- текстовые логи работы агента.

Если пользовательская инструкция предполагает несколько этапов обработки, в решении может дополнительно сохраняться история промежуточных изображений.

## Требования к демонстрации

В решении должна быть продемонстрирована работа агента на нескольких разных сценариях редактирования.

Среди сценариев должны быть примеры задач следующих типов:
- применение фильтров и тоновой коррекции;
- добавление текста на изображение;
- локальное редактирование части изображения;
- удаление или замена объекта;
- использование двух изображений в одном запросе.

- качество итогового редактирования изображений;
- устойчивость к разным формулировкам запросов;
- удобство и полнота логов;
- качество инженерной реализации;
